```python
import gymnasium as gym
import numpy as np
import random
```

```python
env = gym.make('CliffWalking-v1')
```

```python
#q learning define - off policy algo
gamma = 0.99
alpha = 0.5
episodes =  500
epsilon = 0.1 # 10% explore ,90% exploit
```

```python
q = np.zeros((48,4))
```

```python
# epsilon greedy policy used for action but not used in update compute

def epsilon_greedy(state):
  if random.random() < epsilon:
    return env.action_space.sample() # explore done by random action
  else:
    return np.argmax(q[state]) # exploit done
```

```python
for episode in range(episodes):

  render = (episode % 50 == 0) # show learning process of every 50th step = 10 times
  if render:
    env = gym.make('CliffWalking-v1', render_mode="human")
  else:
    env = gym.make('CliffWalking-v1')

  done = False
  state,_ = env.reset() # curr state

  total_reward = 0
  episode_len = 0 # agent moves how many

  while not done: # untill done
    action = epsilon_greedy(state)
    next_state, reward,  terminated,truncated, _ = env.step(action)

    done = terminated or truncated


    q[state, action] +=  alpha * (reward + gamma * np.max(q[next_state] - q[state, action])) # formula

    state = next_state

    total_reward += reward
    episode_len += 1

env.close()
print(f"Total Reward: {total_reward}, Episode Length: {episode_len}")

```
