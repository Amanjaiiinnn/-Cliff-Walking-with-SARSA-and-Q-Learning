```python
import gymnasium as gym
import numpy as np
import random
```

```python
env = gym.make('CliffWalking-v1')
env
```

```python
print(env.action_space.n)
print(env.observation_space.n)

starting_state,_ = env.reset() # 3*12+0
print(starting_state)
```

```python
#sarsa define - on policy algo
gamma = 0.99
alpha = 0.5
episodes =  100   #500
epsilon = 0.1 # 10% explore ,90% exploit
```

```python
# Q table
q = np.zeros((48,4))
q
```

```python
# epsilon greedy policy

def epsilon_greedy(state, q_table):
  if random.random() < epsilon:
    return env.action_space.sample() # explore done by random action
  else:
    return np.argmax(q[state]) # exploit done
```

```python
for episode in range(episodes):

  # show rendering of process but make system slow
  #render = (episode % 50 == 0) # show learning process of every 50th step = 10 times
  # if render:
  #   env = gym.make('CliffWalking-v1', render_mode="human")
  #   print(f"rendering env for episodes={episode+1}/500")
  # else:
  #   env = gym.make('CliffWalking-v1')


  env = gym.make('CliffWalking-v1')

  done = False
  state,_ = env.reset() # curr state
  action = epsilon_greedy(state, q)

  total_reward = 0
  episode_len = 0 # agent moves how many

  while not done: # untill done
    next_state, reward,  terminated,truncated, _ = env.step(action)
    next_action = epsilon_greedy(next_state, q)


    #sarsa update now
    q[state, action] +=  alpha * (reward + gamma * q[next_state, next_action] - q[state, action]) # formula

    state = next_state
    action = next_action

    total_reward += reward
    episode_len += 1

  print(f"Episode: {episode+1}/500, Total Reward: {total_reward}, Episode Length: {episode_len}")
  env.close()
```
